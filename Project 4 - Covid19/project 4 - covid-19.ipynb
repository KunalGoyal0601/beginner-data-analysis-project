```python
import pandas as pd
import seaborn as sns 
import matplotlib.pyplot as plt 
from pathlib import Path

PROJECT_FOLDER = "Project 4 - Covid19"

def load_local_csv(filename):
    """Load a user-supplied dataset; third-party CSVs are not bundled here."""
    candidates = [
        Path.cwd() / filename,
        Path.cwd() / PROJECT_FOLDER / filename,
        Path.cwd().parent / filename,
    ]
    data_path = next((candidate for candidate in candidates if candidate.is_file()), None)
    if data_path is None:
        raise FileNotFoundError(
            f"Missing {filename}. This repository does not redistribute this dataset "
            "because its original source and reuse terms have not been verified. "
            "Obtain it from the original publisher, review the terms, and place it "
            "next to this notebook."
        )
    print(f"Loaded local dataset: {data_path.resolve()}")
    return pd.read_csv(data_path)
```

```python
df = load_local_csv("covid.csv")
```

Q. 1) Show the number of Confirmed, Deaths and Recovered cases in each Region.


```python
df.groupby('Region').sum()
```

Q. 2) Remove all the records where the Confirmed Cases is Less Than 10.


```python
df = df[~(df.Confirmed <10)]
```

```python
df
```

Q. 3) In which Region, maximum number of Confirmed cases were recorded ?


```python
df.groupby('Region').Confirmed.sum().sort_values(ascending = False)
```

Q. 4) In which Region, minimum number of Deaths cases were recorded ?


```python
df.groupby('Region').Deaths.sum().sort_values(ascending = True)
```
