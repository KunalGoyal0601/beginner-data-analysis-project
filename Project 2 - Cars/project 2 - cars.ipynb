```python
import pandas as pd
from pathlib import Path

PROJECT_FOLDER = "Project 2 - Cars"

def load_local_csv(filename):
    """Load a user-supplied dataset; third-party CSVs are not bundled here."""
    candidates = [
        Path.cwd() / filename,
        Path.cwd() / PROJECT_FOLDER / filename,
        Path.cwd().parent / filename,
    ]
    data_path = next((candidate for candidate in candidates if candidate.is_file()), None)
    if data_path is None:
        raise FileNotFoundError(
            f"Missing {filename}. This repository does not redistribute this dataset "
            "because its original source and reuse terms have not been verified. "
            "Obtain it from the original publisher, review the terms, and place it "
            "next to this notebook."
        )
    print(f"Loaded local dataset: {data_path.resolve()}")
    return pd.read_csv(data_path)
```
